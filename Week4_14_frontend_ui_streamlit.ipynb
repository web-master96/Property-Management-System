## Part 2: 💾 Session State (Memory)

### The Problem

Streamlit reruns your entire script on every interaction. Variables get reset!

```python
count = 0  # This resets to 0 every time!
if st.button('Count'):
    count += 1
    st.write(count)  # Always shows 1
```

### The Solution: Session State

`st.session_state` persists data across reruns!

````python
# Session state demo code

SESSION_STATE_APP = '''
import streamlit as st

# Initialize session state
if 'count' not in st.session_state:
    st.session_state.count = 0

if 'messages' not in st.session_state:
    st.session_state.messages = []

st.title('💾 Session State Demo')

# Counter that persists
if st.button('Increment'):
    st.session_state.count += 1

st.write(f'Count: {st.session_state.count}')

# Chat history that persists
user_msg = st.text_input('Your message:')
if st.button('Send') and user_msg:
    st.session_state.messages.append({'role': 'user', 'content': user_msg})
    st.session_state.messages.append({'role': 'bot', 'content': f'Echo: {user_msg}'})

# Display chat history
for msg in st.session_state.messages:
    with st.chat_message(msg['role']):
        st.write(msg['content'])
'''

print('💾 Session State Pattern:')
print(SESSION_STATE_APP)
````

## Part 3: 🎭 Building Your Digital Twin UI

Let's create a complete chat interface for your digital twin!

````python
# Complete Digital Twin Streamlit App

DIGITAL_TWIN_APP = '''
import streamlit as st
from datetime import datetime
import time

# Page config (must be first Streamlit command)
st.set_page_config(
    page_title="Digital Twin Chat",
    page_icon="🤖",
    layout="wide",
    initial_sidebar_state="expanded"
)

# Initialize session state
if 'messages' not in st.session_state:
    st.session_state.messages = []
    # Add welcome message
    st.session_state.messages.append({
        'role': 'assistant',
        'content': 'Hello! I\'m your Digital Twin. How can I help you today?',
        'timestamp': datetime.now()
    })

if 'user_name' not in st.session_state:
    st.session_state.user_name = ''

# Simple response function (replace with your LLM later)
def generate_response(user_message: str) -> str:
    '''Generate a response from the digital twin.'''
    # Placeholder - integrate your Week 1-3 digital twin here!
    responses = {
        'hello': 'Hi there! How are you doing?',
        'help': 'I can chat with you, remember our conversation, and use tools!',
        'time': f'The current time is {datetime.now().strftime("%H:%M:%S")}',
    }

    for key, response in responses.items():
        if key in user_message.lower():
            return response

    return f"You said: {user_message}. I'm learning to respond better!"

# Sidebar
with st.sidebar:
    st.title('⚙️ Settings')

    # User profile
    st.subheader('Profile')
    name = st.text_input('Your Name:', st.session_state.user_name)
    if name != st.session_state.user_name:
        st.session_state.user_name = name

    st.divider()

    # Stats
    st.subheader('📊 Stats')
    st.metric('Messages', len(st.session_state.messages))
    st.metric('User Messages',
              len([m for m in st.session_state.messages if m['role'] == 'user']))

    st.divider()

    # Clear chat
    if st.button('🗑️ Clear Chat', use_container_width=True):
        st.session_state.messages = []
        st.rerun()

# Main chat interface
st.title('🤖 Digital Twin Chat')

if st.session_state.user_name:
    st.write(f'Chatting as: **{st.session_state.user_name}**')

# Chat container
chat_container = st.container()

# Display all messages
with chat_container:
    for message in st.session_state.messages:
        with st.chat_message(message['role']):
            st.write(message['content'])
            if 'timestamp' in message:
                st.caption(message['timestamp'].strftime('%H:%M:%S'))

# Chat input
user_input = st.chat_input('Type your message...')

if user_input:
    # Add user message
    st.session_state.messages.append({
        'role': 'user',
        'content': user_input,
        'timestamp': datetime.now()
    })

    # Display user message immediately
    with chat_container:
        with st.chat_message('user'):
            st.write(user_input)

    # Generate and display response
    with chat_container:
        with st.chat_message('assistant'):
            with st.spinner('Thinking...'):
                time.sleep(0.5)  # Simulate processing
                response = generate_response(user_input)
                st.write(response)

    # Add assistant message to history
    st.session_state.messages.append({
        'role': 'assistant',
        'content': response,
        'timestamp': datetime.now()
    })

    # Force rerun to show new messages
    st.rerun()
'''

print('🎭 Complete Digital Twin Chat UI:')
print('\n💡 Save as digital_twin_app.py and run:')
print('   streamlit run digital_twin_app.py')
print('\n📝 Code preview:')
print(DIGITAL_TWIN_APP[:500] + '...\n[Full code above]')
````

### Exercise 1 (Easy): Basic Streamlit App

**Task**: Create a simple calculator app

**Requirements**:
1. Two number inputs
2. Operation selector (dropdown: +, -, *, /)
3. Calculate button
4. Display result

**Bonus**: Add error handling for division by zero

````python
# Install streamlit in the notebook environment
!pip install -q streamlit

import streamlit as st

# Save this as calculator_app.py
CALCULATOR_APP_CODE = '''
import streamlit as st

st.set_page_config(page_title="Simple Calculator", page_icon="🧮", layout="centered")

st.title(' Simple Calculator')
st.write('Perform basic mathematical operations instantly!')

# Create two columns for number inputs
col1, col2 = st.columns(2)

with col1:
    num1 = st.number_input('Enter first number', value=0.0, step=1.0)

with col2:
    num2 = st.number_input('Enter second number', value=0.0, step=1.0)

# Operation selector
operation = st.selectbox('Choose an operation', ['+', '-', '*', '/'])

# Calculate button
if st.button('Calculate', use_container_width=True):
    with st.spinner('Calculating...'):
        result = None
        error_message = None

        if operation == '+':
            result = num1 + num2
        elif operation == '-':
            result = num1 - num2
        elif operation == '*':
            result = num1 * num2
        elif operation == '/':
            if num2 == 0:
                error_message = "Error: Division by zero is not allowed! "
            else:
                result = num1 / num2

        # Display result or error
        if error_message:
            st.error(error_message)
        else:
            st.success(f'### Result: {result}')
            st.balloons()
'''

# Write the file so the user can easily run it
with open('calculator_app.py', 'w') as f:
    f.write(CALCULATOR_APP_CODE)

print(' Calculator App Code saved to "calculator_app.py"')
print('\n Run this code in your terminal/environment using:')
print('   streamlit run calculator_app.py')
````

### Exercise 2 (Easy): Session State Practice

**Task**: Build a todo list app

**Requirements**:
1. Text input for new todo
2. Add button
3. Display all todos
4. Clear all button
5. Use session state to persist todos

````python
# Save as todo_app.py
TODO_APP_CODE = '''
import streamlit as st

st.set_page_config(page_title="Todo List", page_icon="✅", layout="centered")

st.title('✅ Todo List App')
st.write('Manage and track your tasks effortlessly!')

# Initialize session state for todos
if 'todos' not in st.session_state:
    st.session_state.todos = []

# Form for adding a new todo item
with st.form(key='todo_form', clear_on_submit=True):
    new_todo = st.text_input('What needs to be done?', placeholder='Enter your task here...')
    submit_button = st.form_submit_button(label='Add Task')

# Add the new todo to session state
if submit_button and new_todo.strip() != "":
    st.session_state.todos.append({
        'task': new_todo.strip(),
        'completed': False
    })
    st.success(f'Added task: "{new_todo.strip()}"')

st.divider()

# Display the list of todos with completion checkboxes
if st.session_state.todos:
    st.subheader('Your Tasks')
    for index, todo in enumerate(st.session_state.todos):
        # Use checkboxes to mark tasks as completed
        is_completed = st.checkbox(todo['task'], value=todo['completed'], key=f"todo_{index}")
        st.session_state.todos[index]['completed'] = is_completed
else:
    st.info('Your todo list is currently empty. Add a task above to get started!')

st.divider()

# Clear all button
if st.button('Clear All Tasks', use_container_width=True, type='secondary'):
    st.session_state.todos = []
    st.success('Cleared all tasks!')
    st.rerun()
'''

# Write the file so the user can easily run it
with open('todo_app.py', 'w') as f:
    f.write(TODO_APP_CODE)

print('📝 Todo App Code saved to "todo_app.py"')
print('\n💡 Run this code in your terminal/environment using:')
print('   streamlit run todo_app.py')
````

### Exercise 3 (Intermediate): Chat Interface

**Task**: Build a simple chatbot UI

**Requirements**:
1. Chat input widget
2. Display message history
3. Alternate user/bot messages
4. Add timestamps
5. Clear chat button in sidebar

````python
# Save as chat_app.py
CHAT_APP_CODE = '''
import streamlit as st
from datetime import datetime
import random
import time

st.set_page_config(page_title="Simple Chatbot", page_icon="💬", layout="centered")

# Initialize session state for chat history
if 'messages' not in st.session_state:
    st.session_state.messages = [
        {
            'role': 'assistant',
            'content': 'Hello! How can I assist you today?',
            'timestamp': datetime.now().strftime('%H:%M:%S')
        }
    ]

# Sidebar with configuration and clear history button
with st.sidebar:
    st.title('🤖 Bot Settings')
    st.write('Manage your ongoing conversation session.')
    st.divider()
    if st.button('🧹 Clear Chat History', use_container_width=True):
        st.session_state.messages = [
            {
                'role': 'assistant',
                'content': 'Chat cleared! How can I assist you today?',
                'timestamp': datetime.now().strftime('%H:%M:%S')
            }
        ]
        st.rerun()

# Main chat interface
st.title('💬 Simple Chatbot UI')
st.write('An interactive interface preserving your conversation history.')
st.divider()

# Display conversation history
for message in st.session_state.messages:
    with st.chat_message(message['role']):
        st.write(message['content'])
        st.caption(f"Sent at {message['timestamp']}")

# Chat input box
user_input = st.chat_input('Write a message...')

if user_input:
    # Append user message
    current_time = datetime.now().strftime('%H:%M:%S')
    st.session_state.messages.append({
        'role': 'user',
        'content': user_input,
        'timestamp': current_time
    })

    # Display user message immediately
    with st.chat_message('user'):
        st.write(user_input)
        st.caption(f"Sent at {current_time}")

    # Generate simulated bot response
    with st.chat_message('assistant'):
        with st.spinner('Thinking...'):
            time.sleep(1.0)
            bot_responses = [
                f"That is interesting! Tell me more about: '{user_input}'",
                "I am a simple chatbot interface running live!",
                "Fascinating! How does that relate to your projects?",
                "That sounds great! What is your next plan?"
            ]
            response_content = random.choice(bot_responses)
            response_time = datetime.now().strftime('%H:%M:%S')
            st.write(response_content)
            st.caption(f"Sent at {response_time}")

    # Append bot response to session state
    st.session_state.messages.append({
        'role': 'assistant',
        'content': response_content,
        'timestamp': response_time
    })

    # Rerun to cleanly update components
    st.rerun()
'''

# Write the file so the user can easily run it
with open('chat_app.py', 'w') as f:
    f.write(CHAT_APP_CODE)

print('📝 Chat App Code saved to "chat_app.py"')
print('\n💡 Run this code in your terminal/environment using:')
print('   streamlit run chat_app.py')
````

````python
# Save as digital_twin_advanced.py
DIGITAL_TWIN_ADVANCED_CODE = '''
import streamlit as st
from datetime import datetime
import time
import random
import pandas as pd

st.set_page_config(
    page_title="AI Digital Twin Hub",
    page_icon="🧠",
    layout="wide",
    initial_sidebar_state="expanded"
)

# --- SESSION STATE INITIALIZATION ---
if 'messages' not in st.session_state:
    st.session_state.messages = [
        {
            'role': 'assistant',
            'content': "Hello! I'm your advanced AI Digital Twin. I have access to your customized persona, uploaded knowledge, and external tools. Ask me anything!",
            'timestamp': datetime.now().strftime('%H:%M:%S')
        }
    ]

if 'user_profile' not in st.session_state:
    st.session_state.user_profile = {
        'name': 'Innovator',
        'role': 'Developer',
        'language': 'English'
    }

if 'twin_stats' not in st.session_state:
    st.session_state.twin_stats = {
        'queries_processed': 1,
        'tokens_saved': 1240,
        'knowledge_hits': 0
    }

# --- SIDEBAR: SETTINGS & STATS ---
with st.sidebar:
    st.title("🧠 Twin Control Center")
    st.write("Configure your digital clone's parameters live.")
    st.divider()

    # User Profile Section
    st.subheader("👤 User Profile")
    u_name = st.text_input("Your Name", value=st.session_state.user_profile['name'])
    u_role = st.selectbox("Your Role", ["Developer", "Researcher", "Entrepreneur", "Student"], index=0)
    u_lang = st.selectbox("Primary Language", ["English", "Swahili", "French", "Amharic"], index=0)

    # Save updates back to state
    st.session_state.user_profile['name'] = u_name
    st.session_state.user_profile['role'] = u_role
    st.session_state.user_profile['language'] = u_lang

    st.divider()

    # Model Hyperparameters
    st.subheader("🎛️ Hyperparameters")
    temperature = st.slider("Creativity / Temp", 0.0, 1.0, 0.7, step=0.1, help="Higher values make output more creative")
    active_tools = st.multiselect("Enabled Tools", ["Knowledge Base (RAG)", "Web Search", "Python Executor"], default=["Knowledge Base (RAG)"])

    st.divider()

    # Action Button: Reset Session State
    if st.button("🗑️ Reset Conversations", use_container_width=True, type="secondary"):
        st.session_state.messages = [
            {
                'role': 'assistant',
                'content': f"Reset successful! Hello {u_name}, how can I replicate your thoughts today?",
                'timestamp': datetime.now().strftime('%H:%M:%S')
            }
        ]
        st.session_state.twin_stats['queries_processed'] = 1
        st.session_state.twin_stats['knowledge_hits'] = 0
        st.success("Memory cleared!")
        st.rerun()

# --- MAIN PAGE LAYOUT ---
st.title("🚀 Complete Digital Twin Core")
st.write(f"Welcome back, **{st.session_state.user_profile['name']}** ({st.session_state.user_profile['role']}). System language is active in **{st.session_state.user_profile['language']}**.")

# Real-time Metrics Dashboard Row
metric_col1, metric_col2, metric_col3 = st.columns(3)
with metric_col1:
    st.metric(label="Total Exchanges", value=st.session_state.twin_stats['queries_processed'], delta="+1 new")
with metric_col2:
    st.metric(label="RAG Knowledge Hits", value=st.session_state.twin_stats['knowledge_hits'], delta="+1 dynamic")
with metric_col3:
    st.metric(label="Est. Time Saved (mins)", value=round(st.session_state.twin_stats['queries_processed'] * 2.5, 1))

st.divider()

# Tabs for Chat vs. Analytics Logs
tab_chat, tab_analytics = st.tabs(["💬 Dynamic Chat Arena", "📊 Diagnostic Logs"])

with tab_chat:
    st.subheader("Interactive Persona Emulation")

    # Render conversation history inside a container
    chat_container = st.container()
    with chat_container:
        for msg in st.session_state.messages:
            with st.chat_message(msg['role']):
                st.markdown(f"**{msg['content']}**")
                st.caption(f"🕒 {msg['timestamp']}")

    # Accept query inputs
    user_query = st.chat_input("Ask your digital twin a question...")

    if user_query:
        # Time tracking and appending user text
        u_time = datetime.now().strftime('%H:%M:%S')
        st.session_state.messages.append({
            'role': 'user',
            'content': user_query,
            'timestamp': u_time
        })

        # Force display immediately
        with chat_container:
            with st.chat_message('user'):
                st.markdown(f"**{user_query}**")
                st.caption(f"🕒 {u_time}")

        # Simulated AI inference & tool calling
        with chat_container:
            with st.chat_message('assistant'):
                with st.spinner('Accessing Knowledge Bases & Tools...'):
                    time.sleep(1.2)

                    # Simulated response strategy
                    twin_response = ""
                    if "rag" in [t.lower() for t in active_tools] or "knowledge" in user_query.lower():
                        st.session_state.twin_stats['knowledge_hits'] += 1
                        twin_response += "[📚 Knowledge Source Active] From our internal documentation repository: "

                    responses_pool = [
                        f"Based on your preferred setup ({temperature} temp), I recommend automating this workflow directly in Python.",
                        f"That aligns with your goals as a {u_role}! Here is how you usually frame this scenario.",
                        f"This matches our previous projects. Let me know if you want me to write code to test this!",
                        f"Let's tackle this from an engineering perspective. What are the main blockers?"
                    ]

                    twin_response += random.choice(responses_pool)
                    bot_time = datetime.now().strftime('%H:%M:%S')
                    st.markdown(twin_response)
                    st.caption(f"🕒 {bot_time}")

        # Save back state
        st.session_state.messages.append({
            'role': 'assistant',
            'content': twin_response,
            'timestamp': bot_time
        })
        st.session_state.twin_stats['queries_processed'] += 1
        st.rerun()

with tab_analytics:
    st.subheader("System Activity Logs")
    st.write("Below is a diagnostic feed tracking session updates, token savings, and state tracking.")

    log_data = pd.DataFrame({
        'Timestamp': [datetime.now().strftime('%Y-%m-%d %H:%M:%S')],
        'Client Name': [st.session_state.user_profile['name']],
        'Current Role': [st.session_state.user_profile['role']],
        'Active Language': [st.session_state.user_profile['language']],
        'Total Exchange Count': [st.session_state.twin_stats['queries_processed']]
    })
    st.dataframe(log_data, use_container_width=True)

    with st.expander("🛠️ View Raw System State Reference"):
        st.write(st.session_state.to_dict())
'''

# Save the app file locally
with open('digital_twin_advanced.py', 'w') as f:
    f.write(DIGITAL_TWIN_ADVANCED_CODE)

print('🏆 Advanced Digital Twin UI saved to "digital_twin_advanced.py"')
print('\n💡 Run your completed production interface via:')
print('   streamlit run digital_twin_advanced.py')
````
