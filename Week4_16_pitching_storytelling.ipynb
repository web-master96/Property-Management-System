## Part 1: 📐 The Problem-Solution-Impact Framework

### The Y-Combinator Formula

````python
# The PSI Framework

PSI_FRAMEWORK = {
    'Problem (30 seconds)': {
        'What': 'The pain point you\'re solving',
        'Questions to answer': [
            'Who has this problem?',
            'How big is the problem?',
            'Why does it matter?',
            'What happens if unsolved?',
        ],
        'Common mistakes': [
            '❌ Starting with your solution',
            '❌ Using jargon',
            '❌ Making problem too small',
            '❌ Not showing urgency',
        ],
        'Good example': [
            '"In rural Kenya, 60% of farmers lose 30% of their crops',
            'to diseases they could have prevented. By the time they',
            'realize what\'s wrong, it\'s too late. This costs them',
            '$200/year - 20% of their income."',
        ],
    },
    'Solution (45 seconds)': {
        'What': 'How you solve the problem',
        'Questions to answer': [
            'What does your product do?',
            'How does it work? (high-level)',
            'Why is it better than alternatives?',
            'What makes it feasible now?',
        ],
        'Common mistakes': [
            '❌ Too much technical detail',
            '❌ Feature list instead of benefits',
            '❌ Not showing differentiation',
            '❌ Ignoring existing solutions',
        ],
        'Good example': [
            '"We built an AI assistant that identifies crop diseases',
            'via WhatsApp. Farmers take a photo, get diagnosis in 30',
            'seconds, in their language. Unlike extension officers who',
            'visit monthly, we\'re available 24/7 for free."',
        ],
    },
    'Impact (45 seconds)': {
        'What': 'The results and potential',
        'Questions to answer': [
            'Who have you helped so far?',
            'What were the results?',
            'What\'s the potential scale?',
            'Why will this succeed?',
        ],
        'Common mistakes': [
            '❌ Only talking about potential',
            '❌ No concrete numbers',
            '❌ Unrealistic projections',
            '❌ Not showing traction',
        ],
        'Good example': [
            '"We\'ve helped 500 farmers in 3 months. They\'ve saved',
            'an average of $150 this season. 89% say they\'ll keep using',
            'it. There are 5 million farmers in Kenya alone. If we help',
            '1% of them, that\'s $7.5M in saved income annually."',
        ],
    },
}

print('📐 Problem-Solution-Impact Framework:\n')
for section, details in PSI_FRAMEWORK.items():
    print(f'\n{section}:')
    print(f"  {details['What']}\n")
    print('  Questions:')
    for q in details['Questions to answer']:
        print(f'    • {q}')
    print('\n  Common Mistakes:')
    for m in details['Common mistakes']:
        print(f'    {m}')
    print('\n  Example:')
    for line in details['Good example']:
        print(f'    {line}')
    print()
````

### Building Your Pitch: Step by Step

````python
# Pitch builder template

class PitchBuilder:
    '''Helper class to structure your pitch.'''

    def __init__(self):
        self.pitch = {
            'problem': {
                'who': '',
                'what': '',
                'size': '',
                'urgency': '',
            },
            'solution': {
                'what': '',
                'how': '',
                'differentiation': '',
                'why_now': '',
            },
            'impact': {
                'traction': '',
                'results': '',
                'potential': '',
                'why_win': '',
            }
        }

    def add_problem(self, who, what, size, urgency):
        '''Define the problem.'''
        self.pitch['problem'] = {
            'who': who,
            'what': what,
            'size': size,
            'urgency': urgency,
        }
        return self

    def add_solution(self, what, how, differentiation, why_now):
        '''Define your solution.'''
        self.pitch['solution'] = {
            'what': what,
            'how': how,
            'differentiation': differentiation,
            'why_now': why_now,
        }
        return self

    def add_impact(self, traction, results, potential, why_win):
        '''Define the impact.'''
        self.pitch['impact'] = {
            'traction': traction,
            'results': results,
            'potential': potential,
            'why_win': why_win,
        }
        return self

    def generate_script(self):
        '''Generate a pitch script.'''
        script = []

        # Problem section
        script.append('=== PROBLEM (30s) ===')
        p = self.pitch['problem']
        script.append(f"{p['who']} face {p['what']}.")
        script.append(f"This affects {p['size']}.")
        script.append(f"{p['urgency']}")
        script.append('')

        # Solution section
        script.append('=== SOLUTION (45s) ===')
        s = self.pitch['solution']
        script.append(f"We built {s['what']}.")
        script.append(f"It works by {s['how']}.")
        script.append(f"Unlike alternatives, {s['differentiation']}.")
        script.append(f"This is possible now because {s['why_now']}.")
        script.append('')

        # Impact section
        script.append('=== IMPACT (45s) ===')
        i = self.pitch['impact']
        script.append(f"So far, {i['traction']}.")
        script.append(f"The results: {i['results']}.")
        script.append(f"The potential: {i['potential']}.")
        script.append(f"We'll win because {i['why_win']}.")

        return '\n'.join(script)

    def estimate_time(self):
        '''Estimate speaking time (rough).'''
        script = self.generate_script()
        words = len(script.split())
        # Average speaking rate: 150 words/minute
        minutes = words / 150
        return f"{minutes:.1f} minutes ({words} words)"

# Example usage
pitch = PitchBuilder()
pitch.add_problem(
    who="Rural farmers in Kenya",
    what="crop diseases they can't identify",
    size="5 million farmers, losing 30% of crops",
    urgency="This costs them $200/year - 20% of income"
)
pitch.add_solution(
    what="an AI crop disease detector via WhatsApp",
    how="farmers snap a photo and get instant diagnosis in Swahili",
    differentiation="we're available 24/7 while extension officers visit monthly",
    why_now="smartphones are now in 80% of rural areas, and AI models are accessible"
)
pitch.add_impact(
    traction="we've helped 500 farmers in 3 months",
    results="they saved $150 on average this season, 89% retention",
    potential="5M farmers in Kenya, 300M across Africa",
    why_win="we're the only solution in their language, on their platform"
)

print('📝 Example Pitch:\n')
print(pitch.generate_script())
print(f'\nEstimated time: {pitch.estimate_time()}')
````

## Part 2: 📖 Storytelling Techniques

### Why Stories Matter

````python
# The power of storytelling

STORYTELLING_PRINCIPLES = {
    '1. Start with a person': {
        'Why': 'Humans connect with humans, not statistics',
        'Bad': 'Healthcare costs $3.5 trillion annually',
        'Good': 'Meet Sarah. She\'s a single mom who can\'t afford her insulin.',
    },
    '2. Show the struggle': {
        'Why': 'Tension creates engagement',
        'Bad': 'People have trouble finding doctors',
        'Good': 'Sarah waited 3 weeks for an appointment. By then, it was too late.',
    },
    '3. Make it specific': {
        'Why': 'Details make stories real',
        'Bad': 'We help many farmers',
        'Good': 'Last Tuesday, Joseph saved his maize crop. He\'s now $300 richer.',
    },
    '4. Show transformation': {
        'Why': 'People want to see change',
        'Bad': 'Our app is useful',
        'Good': 'Before: Joseph lost 40% of crops. After: He caught disease early, saved it all.',
    },
    '5. Use emotion': {
        'Why': 'Decisions are emotional',
        'Bad': 'Revenue increased 30%',
        'Good': 'When Joseph saw he could pay school fees, he cried. So did we.',
    },
    '6. End with hope': {
        'Why': 'People invest in positive futures',
        'Bad': 'We need funding',
        'Good': 'Imagine 5 million Josephs sending their kids to school. We can make it happen.',
    },
}

print('📖 Storytelling Principles:\n')
for principle, details in STORYTELLING_PRINCIPLES.items():
    print(f'{principle}:')
    print(f"  Why: {details['Why']}")
    print(f"  ❌ Bad: {details['Bad']}")
    print(f"  ✅ Good: {details['Good']}")
    print()
````

````python
# Demo guidelines

DEMO_PRINCIPLES = {
    'Preparation': [
        '✅ Test your demo 10 times',
        '✅ Have backup (video recording)',
        '✅ Pre-load accounts/data',
        '✅ Close unnecessary tabs/apps',
        '✅ Full screen mode',
        '✅ Zoom/font size for visibility',
        '✅ Turn off notifications',
        '✅ Check internet connection',
    ],
    'During Demo': [
        '✅ Start with the problem (again)',
        '✅ Show the "before" state',
        '✅ Walk through ONE complete flow',
        '✅ Narrate what you\'re doing',
        '✅ Highlight key features only',
        '✅ Show real data if possible',
        '✅ End with results',
        '❌ Don\'t show every feature',
        '❌ Don\'t fix bugs during demo',
        '❌ Don\'t apologize for UI',
    ],
    'Common Mistakes': [
        '❌ Demoing features nobody asked for',
        '❌ Getting lost in your own app',
        '❌ Technical difficulties (no backup)',
        '❌ Talking too fast',
        '❌ Showing too much',
        '❌ Not setting context',
        '❌ Forgetting to show impact',
    ],
}

print('🎬 Demo Best Practices:\n')
for category, items in DEMO_PRINCIPLES.items():
    print(f'{category}:')
    for item in items:
        print(f'  {item}')
    print()
````

## Part 4: ❓ Handling Questions

### The Q&A Framework

````python
# Q&A techniques

QA_FRAMEWORK = {
    'Active Listening': {
        'Do': [
            '✅ Listen to the full question',
            '✅ Pause before answering',
            '✅ Repeat question if unclear',
            '✅ Thank them for asking',
        ],
        'Don\'t': [
            '❌ Interrupt',
            '❌ Get defensive',
            '❌ Assume you know the question',
        ]
    },
    'Bridging Technique': {
        'Definition': 'Connect their question to your key points',
        'Example': [
            'Q: "What if farmers don\'t have smartphones?"',
            'A: "Great question. That\'s actually why we chose WhatsApp - ',
            '    it runs on basic phones and uses minimal data. But more ',
            '    importantly, 80% of our target farmers already have ',
            '    smartphones, and that number is growing 20% annually."',
        ]
    },
    'Tough Questions': {
        'Competition': {
            'Q': '"What about [competitor]?"',
            'A': '"They\'re great for X. We focus on Y, which matters more ',
                 'because Z. Here\'s our differentiation: [key unique value]."',
        },
        'Criticism': {
            'Q': '"This will never scale"',
            'A': '"I appreciate the concern. Let me share why we believe it will: ',
                 '[evidence, traction, plan]. What specific scaling concern do you have?"',
        },
        'Doubt': {
            'Q': '"Will people actually use this?"',
            'A': '"That\'s the key question. Here\'s what we\'ve learned: [user data, ',
                 'retention, testimonials]. Would you like to talk to our users?"',
        },
        'Don\'t Know': {
            'Q': '"What\'s your CAC/LTV ratio?"',
            'A': '"That\'s a great question. I don\'t have that exact number with me, ',
                 'but let me get back to you with it. What I can tell you is [related info]."',
        },
    },
    'Red Flags': [
        '🚩 Getting defensive',
        '🚩 Rambling',
        '🚩 Making up answers',
        '🚩 Criticizing the question',
        '🚩 Ignoring the question',
    ],
}

print('❓ Q&A Framework:\n')
print('Active Listening:')
for do in QA_FRAMEWORK['Active Listening']['Do']:
    print(f'  {do}')
for dont in QA_FRAMEWORK['Active Listening']["Don't"]:
    print(f'  {dont}')
print('\nBridging Example:')
for line in QA_FRAMEWORK['Bridging Technique']['Example']:
    print(f'  {line}')
print('\nHandling Tough Questions:')
for q_type, qa in QA_FRAMEWORK['Tough Questions'].items():
    print(f'\n  {q_type}:')
    print(f"    Q: {qa['Q']}")
    print(f"    A: {qa['A']}")
print('\nRed Flags to Avoid:')
for flag in QA_FRAMEWORK['Red Flags']:
    print(f'  {flag}')
````

## Part 5: 🎤 Presentation Skills

### Delivery Techniques

````python
# Presentation delivery tips

DELIVERY_TIPS = {
    'Voice': {
        'Pace': 'Speak slower than you think - pause for emphasis',
        'Volume': 'Project confidence, not just volume',
        'Tone': 'Vary your tone - monotone kills attention',
        'Fillers': 'Eliminate "um", "uh", "like", "you know"',
        'Exercise': 'Record yourself, identify patterns, practice',
    },
    'Body Language': {
        'Posture': 'Stand tall, shoulders back, feet shoulder-width',
        'Hands': 'Use natural gestures, avoid pockets/crossed arms',
        'Eyes': 'Make eye contact with different audience members',
        'Movement': 'Walk with purpose, don\'t pace nervously',
        'Face': 'Smile naturally, show emotion',
    },
    'Slides (if using)': {
        'Rule of thumb': '1 slide = 1 minute max',
        'Content': 'Images > text, headlines only',
        'Font': 'Size 30+ minimum, readable from back',
        'Colors': 'High contrast, consistent theme',
        'Animations': 'Minimal or none',
        'Best practice': 'You are the presentation, not the slides',
    },
    'Energy Management': {
        'Start': 'High energy, confident opening',
        'Middle': 'Maintain enthusiasm, vary tone',
        'End': 'Strong close, memorable line',
        'Nerves': 'Channel as excitement, take deep breaths',
    },
}

print('🎤 Presentation Delivery:\n')
for category, tips in DELIVERY_TIPS.items():
    print(f'{category}:')
    if isinstance(tips, dict):
        for tip_name, tip_value in tips.items():
            print(f'  • {tip_name}: {tip_value}')
    else:
        print(f'  {tips}')
    print()
````

3. 🎬 Demo script with narration
4. 🎤 Video recording of full presentation
5. 📊 3-5 slide deck (optional)
6. ❓ Q&A preparation (10 likely questions)
7. 📈 One-pager summary

**Quality Standards**:
- ✅ Clear, jargon-free language
- ✅ Compelling storytelling
- ✅ Strong evidence of traction
- ✅ Professional delivery
- ✅ Confident Q&A responses
- ✅ Under 3 minutes total

**Test Your Pitch**:

1. Someone in tech (for technical accuracy)
2. Someone not in tech (for clarity)
3. Potential user (for relevance)
4. Potential investor/supporter (for compelling-ness)

**Success Criteria**:
- ✅ 4/5 rating on feedback form (average)
- ✅ People remember your pitch 24 hours later
- ✅ At least one person asks follow-up questions
- ✅ You feel confident presenting it
