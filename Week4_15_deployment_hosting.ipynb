## Part 2: 🔐 Managing Secrets

**Never commit API keys to git!**

````python
# Secure secrets management

SECRETS_PATTERN = '''
# In your app.py
import streamlit as st
import os

# Method 1: Streamlit secrets (recommended for deployment)
def get_api_key():
    """Get API key from Streamlit secrets or user input."""
    # Try to get from Streamlit secrets first
    try:
        api_key = st.secrets["ANTHROPIC_API_KEY"]
        return api_key
    except (FileNotFoundError, KeyError):
        # Fall back to user input
        st.sidebar.warning("⚠️ API key not configured")
        api_key = st.sidebar.text_input(
            "Enter your Anthropic API key:",
            type="password",
            help="Get your key from console.anthropic.com"
        )
        if not api_key:
            st.error("Please provide an API key to continue")
            st.stop()
        return api_key

# Method 2: Environment variables (for local development)
from dotenv import load_dotenv
load_dotenv()  # Loads from .env file

api_key = os.getenv("ANTHROPIC_API_KEY")
'''

print('🔐 Secrets Management Pattern:')
print(SECRETS_PATTERN)
print('\n📝 For Hugging Face Spaces:')
print('  1. Go to Space settings')
print('  2. Add secrets in "Repository secrets" section')
print('  3. Access via st.secrets["KEY_NAME"]')
````

### Exercise 2 (Intermediate): Deploy With Secrets

**Task**: Deploy an app that uses API keys

**Requirements**:
1. App uses Anthropic API
2. Secrets managed securely
3. Fallback to user input if no secret
4. Test both methods
5. Document setup in README

````python
anthropic_app_code = '''
import streamlit as st
from anthropic import Anthropic

st.title("🤖 Secure AI Assistant")
st.write("This assistant uses the Anthropic Claude API to generate responses. Secrets are managed securely.")

# Secure secrets management with fallback
def get_api_key():
    # Try to retrieve from Streamlit secrets
    try:
        return st.secrets["ANTHROPIC_API_KEY"]
    except (FileNotFoundError, KeyError):
        # Fallback to sidebar user input
        st.sidebar.warning("⚠️ ANTHROPIC_API_KEY secret not found in Space settings.")
        user_key = st.sidebar.text_input(
            "Enter your Anthropic API Key:",
            type="password",
            help="Get a key from console.anthropic.com"
        )
        if not user_key:
            st.info("Please enter your Anthropic API Key in the sidebar to proceed.")
            st.stop()
        return user_key

api_key = get_api_key()

# Initialize client
client = Anthropic(api_key=api_key)

# Message input
user_prompt = st.text_input("Ask Claude anything:", placeholder="How does cloud deployment work?")

if user_prompt:
    with st.spinner("🧠 Thinking..."):
        try:
            message = client.messages.create(
                model="claude-3-haiku-20240307",
                max_tokens=1000,
                temperature=0.7,
                messages=[
                    {"role": "user", "content": user_prompt}
                ]
            )
            st.subheader("📝 Response:")
            st.write(message.content[0].text)
        except Exception as e:
            st.error(f"An error occurred: {str(e)}")
'''

# Save app.py
with open('app.py', 'w') as f:
    f.write(anthropic_app_code)

# Save requirements.txt
with open('requirements.txt', 'w') as f:
    f.write('streamlit==1.29.0\nanthropic==0.8.0\n')

# Save README.md
readme_secrets_content = '''---
title: Secure AI Assistant
emoji: 🤖
colorFrom: red
colorTo: yellow
sdk: streamlit
sdk_version: 1.29.0
app_file: app.py
pinned: false
---

# Secure AI Assistant
An AI-powered application utilizing Anthropic's Claude model with secure secret handling.

## Setup and Secrets Configuration
1. Deploy this Space on Hugging Face.
2. Navigate to **Settings** -> **Repository secrets**.
3. Add a new secret with the name `ANTHROPIC_API_KEY` and input your API key from console.anthropic.com.
4. If the key is not configured, the app will gracefully fall back to a secure user input field in the sidebar.
'''

with open('README.md', 'w') as f:
    f.write(readme_secrets_content)

print('✅ Overwrote app.py with AI Assistant code.')
print('✅ Updated requirements.txt with anthropic dependency.')
print('✅ Overwrote README.md with configuration details.')
````

### Exercise 3 (Intermediate): Performance Optimization

**Task**: Optimize a slow app

**Given**: App that loads data on every run

**Requirements**:
1. Add @st.cache_data where appropriate
2. Implement pagination
3. Add loading indicators
4. Measure before/after performance
5. Document improvements

````python
# YOUR CODE HERE
# Optimize the slow app
````

````python
optimized_app_code = '''
import streamlit as st
import pandas as pd
import numpy as np
import time

st.title("⚡ Performance Optimized App")
st.write("This app demonstrates performance optimizations using caching and pagination to load large data efficiently.")

# --- Unoptimized Data Load (Simulated) ---
def load_data_slow():
    time.sleep(2)  # Simulate network latency or heavy computing
    return pd.DataFrame({
        'ID': range(1, 1001),
        'Value': np.random.randn(1000),
        'Category': np.random.choice(['A', 'B', 'C', 'D'], size=1000)
    })

# --- Optimized Cached Data Load ---
@st.cache_data(show_spinner=False)
def load_data_cached():
    time.sleep(2)  # Simulated latency (only runs once on first execution)
    return pd.DataFrame({
        'ID': range(1, 1001),
        'Value': np.random.randn(1000),
        'Category': np.random.choice(['A', 'B', 'C', 'D'], size=1000)
    })

# Performance Measurement Toggle
use_cache = st.sidebar.checkbox("Enable @st.cache_data (Optimization)", value=True)

# Measure execution time
start_time = time.time()

with st.spinner("📥 Loading dataset..."):
    if use_cache:
        df = load_data_cached()
    else:
        df = load_data_slow()

end_time = time.time()
elapsed_time = end_time - start_time

# Display Performance Metrics
st.metric(label="Data Loading Duration", value=f"{elapsed_time:.4f} seconds",
          delta=f"-{2.0 - elapsed_time:.4f}s saved" if use_cache and elapsed_time < 0.1 else None)

# --- Implementation of Pagination ---
st.subheader("📊 Dataset Viewer (Paginated)")

# User settings for pagination
page_size = st.sidebar.slider("Rows per page", min_value=5, max_value=50, value=10)
total_rows = len(df)
total_pages = int(np.ceil(total_rows / page_size))

current_page = st.number_input("Page", min_value=1, max_value=total_pages, value=1, step=1)

# Calculate start and end indices
start_idx = (current_page - 1) * page_size
end_idx = min(start_idx + page_size, total_rows)

# Display paginated data subset
st.dataframe(df.iloc[start_idx:end_idx], use_container_width=True)
st.write(f"Showing rows {start_idx + 1} to {end_idx} of {total_rows} entries.")
'''

# Save optimized app.py
with open('app.py', 'w') as f:
    f.write(optimized_app_code)

# Update requirements.txt
with open('requirements.txt', 'w') as f:
    f.write('streamlit==1.29.0\npandas==2.1.4\nnumpy==1.26.2\n')

# Update README.md
readme_opt_content = '''---
title: Performance Optimized App
emoji: ⚡
colorFrom: purple
colorTo: pink
sdk: streamlit
sdk_version: 1.29.0
app_file: app.py
pinned: false
---

# Performance Optimized App

This application resolves bottlenecks associated with heavy analytical operations or databases by implementing `@st.cache_data` and client-side pagination.

## Optimizations Implemented
- **@st.cache_data**: Avoids re-generating or querying the 1,000-row dataset on every interactive slider or pagination change.
- **Pagination**: Restricts UI render load to manageable chunks (e.g., 10 rows), preventing sluggish scroll performance on low-bandwidth networks.
- **Interactive Performance Monitor**: Live benchmark tracking on top of the dashboard.
'''

with open('README.md', 'w') as f:
    f.write(readme_opt_content)

print('✅ Generated optimized app.py')
print('✅ Updated requirements.txt with pandas & numpy')
print('✅ Updated README.md with optimization details')
````
